```python
import pandas as pd

# Load the uploaded CSV file
file_path_test = "E:/CAP_FPT/2. CODE/2. Module 1 design/1. Hoang van Bac/ANLALYSIT_BB_UD_RESULT/MODEL_UD_ORIGIN/metrics/test/20250519_230738/all_metrics_summary.csv"
file_path_train_valid = "E:/CAP_FPT/2. CODE/2. Module 1 design/1. Hoang van Bac/ANLALYSIT_BB_UD_RESULT/MODEL_UD_ORIGIN/metrics/training/20250514_155337/all_metrics_summary.csv"

df_test = pd.read_csv(file_path_test)
df_train_valid = pd.read_csv(file_path_train_valid)
```

```python
# Kiểm tra xem tập train_valid có đủ các loại Dataset không
print("Các giá trị Dataset trong df_train_valid:", df_train_valid['Dataset'].unique())
```

```python
# Tính trung bình theo từng mô hình cho từng loại Dataset
mean_train = df_train_valid[df_train_valid['Dataset'] == 'train'].groupby('Model').mean(numeric_only=True).reset_index()
mean_valid = df_train_valid[df_train_valid['Dataset'] == 'valid'].groupby('Model').mean(numeric_only=True).reset_index()
mean_test = df_test.groupby('Model').mean(numeric_only=True).reset_index()
```

```python

# Hiển thị kết quả
print("Trung bình TRAIN:")
print(mean_train)

print("\nTrung bình VALID:")
print(mean_valid)

print("\nTrung bình TEST:")
print(mean_test)
```
